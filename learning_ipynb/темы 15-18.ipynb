```python
#Lambda-функции, это способ создать маленькую, безымянную функцию в одну строку, обычно когда функция нужна лишь один раз, "на лету" — например, для сортировки по особому критерию.
# Обычная функция
def square(x):
    return x * x

# То же самое через lambda
square = lambda x: x * x

print(square(5))  # 25 — работает одинаково

#Реальное применение — как аргумент для sorted(), когда нужно отсортировать по нестандартному критерию:
drinks = [("Латте", 200), ("Эспрессо", 120), ("Раф", 250)]

sorted_by_price = sorted(drinks, key=lambda item: item[1])
print(sorted_by_price)  # [('Эспрессо', 120), ('Латте', 200), ('Раф', 250)]
#Тут lambda item: item[1] говорит sorted(): "сортируй, глядя именно на второй элемент каждого кортежа (цену)".
```

```python
#Генераторы списков (list comprehensions)
# Обычный способ
squares = []
for x in range(1, 6):
    squares.append(x * x)

# То же самое через генератор списка
squares = [x * x for x in range(1, 6)]
print(squares)  # [1, 4, 9, 16, 25]
#С условием (фильтрация):
# Только чётные квадраты
squares = [x * x for x in range(1, 11) if x % 2 == 0]
print(squares)  # [4, 16, 36, 64, 100]
```
